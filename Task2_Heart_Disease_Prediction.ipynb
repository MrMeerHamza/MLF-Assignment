### Step 2: Load Raw Data

````python
"""
Purpose of this section:
-------------------------
This part of the code loads the Cleveland heart disease data (downloaded from UCI)
into a pandas DataFrame. The file has no header row, so the 14 attribute names
from heart-disease.names are assigned. Missing values are written as '?' in the file.
"""

attribute_names = ["age", "sex", "cp", "trestbps", "chol", "fbs", "restecg",
                   "thalach", "exang", "oldpeak", "slope", "ca", "thal", "num"]

# Read the data file; '?' marks a missing value
raw_data = pd.read_csv(os.path.join(PROJECT_DIR, "processed.cleveland.data"),
                       header=None, names=attribute_names, na_values="?")

print("\n\nRaw Data (first 15 rows):")
print("=========================\n")
pd.set_option("display.max_columns", None, "display.width", 200)
print(raw_data.head(15))
````

### Step 3: Understand and Pre-process Data

#### Step 3.1: Understand Raw Data

````python
"""
Purpose of this section:
-------------------------
This part of the code helps us understand the dataset by:
1. Printing the names of all attributes (columns) in the DataFrame.
2. Showing the total number of instances (rows) available in the dataset.
3. Showing how many values are missing in each attribute.
4. Showing the distribution of the output attribute (num).
"""

print("\n\nAttributes in Raw Data:")
print("=======================\n")
print(raw_data.columns)

print("\n\nNumber of Instances in Raw Data:", raw_data["age"].count())
print("====================================\n")

print("Missing values per attribute:")
print(raw_data.isna().sum()[raw_data.isna().sum() > 0])

print("\nOutput attribute 'num' (0 = no disease, 1-4 = disease):")
print(raw_data["num"].value_counts().sort_index())
````

#### Step 3.2: Pre-process Raw Data

o   **Missing values:** `ca` has 4 and `thal` has 2 missing values (6 rows in total). Since this is only 2% of the data, these rows are removed.

o   **Output attribute:** `num` has values 0 – 4. Values 1, 2, 3 and 4 all mean heart disease is present, so the output is converted into
    a **Yes / No** attribute named `HeartDisease` (0 → No, 1 – 4 → Yes), just like `Survived` in the Titanic system.

````python
"""
Purpose of this section:
-------------------------
This part of the code cleans the raw data:
1. Removes the rows that contain missing values.
2. Converts the output attribute 'num' (0-4) into 'HeartDisease' (No / Yes).
"""

# Remove rows with missing values
preprocessed_data = raw_data.dropna().copy()
print("Rows before removing missing values:", len(raw_data))
print("Rows after removing missing values :", len(preprocessed_data))

# Codes of categorical attributes are whole numbers (stored as 1.0, 3.0, ... in the file)
for attribute in ["sex", "cp", "fbs", "restecg", "exang", "slope", "ca", "thal"]:
    preprocessed_data[attribute] = preprocessed_data[attribute].astype(int)

# Convert output: 0 -> "No", 1-4 -> "Yes"
preprocessed_data["HeartDisease"] = np.where(preprocessed_data["num"] > 0, "Yes", "No")
preprocessed_data = preprocessed_data.drop(columns=["num"])

print("\nHeartDisease (class balance):")
print(preprocessed_data["HeartDisease"].value_counts())
````

### Step 4: Feature Selection and Extraction

The assignment allows **only 4 input attributes**. The Titanic system uses **categorical** input attributes that are label encoded,
so to follow exactly the same steps, the 4 attributes are chosen from the 8 categorical attributes of the dataset
(sex, cp, fbs, restecg, exang, slope, ca, thal).

#### Step 4.1: Rank the Categorical Attributes (Cramér's V)

**Cramér's V** measures how strongly a categorical attribute is associated with the output (0 = no association, 1 = perfect association).
It is calculated from the chi-square test on the attribute-vs-output contingency table.

````python
"""
Purpose of this section:
-------------------------
This part of the code calculates Cramér's V between every categorical attribute
and the output (HeartDisease), and ranks the attributes from strongest to weakest.
"""

def cramers_v(attribute_values, output_values):
    contingency_table = pd.crosstab(attribute_values, output_values)
    chi2 = chi2_contingency(contingency_table)[0]
    n = contingency_table.values.sum()
    rows, columns = contingency_table.shape
    return np.sqrt(chi2 / (n * (min(rows, columns) - 1)))

categorical_attributes = ["sex", "cp", "fbs", "restecg", "exang", "slope", "ca", "thal"]
attribute_ranking = pd.Series(
    {attribute: cramers_v(preprocessed_data[attribute], preprocessed_data["HeartDisease"]) for attribute in categorical_attributes}
).sort_values(ascending=False)

ranking_table = PrettyTable(["Rank", "Attribute", "Cramér's V"])
for rank, (attribute, value) in enumerate(attribute_ranking.items(), 1):
    ranking_table.add_row([rank, attribute, round(value, 3)])
print(ranking_table)

plt.figure(figsize=(7, 4))
sns.barplot(x=attribute_ranking.index, y=attribute_ranking.values, hue=attribute_ranking.index, legend=False, palette="Blues_r")
plt.ylabel("Cramér's V with HeartDisease")
plt.xlabel("Categorical attribute")
plt.title("Association of Each Attribute with Heart Disease")
plt.tight_layout()
plt.show()
````

#### Step 4.2: Select the Top 4 Attributes and Create the Sample Data

The 4 strongest attributes are **thal**, **cp**, **ca** and **exang**. Their numeric codes are converted into readable text values
(using `heart-disease.names`) so the sample data looks exactly like the Titanic `sample-data.csv`: text inputs and a Yes/No output.

````python
"""
Purpose of this section:
-------------------------
This part of the code:
1. Selects the top 4 attributes from the ranking in Step 4.1.
2. Converts their numeric codes into readable text values.
3. Saves the result as sample-data.csv (same format as the Titanic sample data).
"""

selected_attributes = list(attribute_ranking.index[:4])
print("Top 4 attributes:", selected_attributes)
assert set(selected_attributes) == {"thal", "cp", "ca", "exang"}

# Code -> text value for each selected attribute (from heart-disease.names)
chest_pain_values = {1: "Typical Angina", 2: "Atypical Angina", 3: "Non-Anginal Pain", 4: "Asymptomatic"}
exercise_angina_values = {0: "No", 1: "Yes"}
major_vessels_values = {0: "Zero", 1: "One", 2: "Two", 3: "Three"}
thal_values = {3: "Normal", 6: "Fixed Defect", 7: "Reversible Defect"}

sample_data = pd.DataFrame({
    "ChestPainType": preprocessed_data["cp"].map(chest_pain_values),
    "ExerciseAngina": preprocessed_data["exang"].map(exercise_angina_values),
    "MajorVessels": preprocessed_data["ca"].map(major_vessels_values),
    "Thal": preprocessed_data["thal"].map(thal_values),
    "HeartDisease": preprocessed_data["HeartDisease"],
}).reset_index(drop=True)

sample_data.to_csv(os.path.join(PROJECT_DIR, "sample-data.csv"), index=False, header=True)

print("\n\nSample Data (first 15 rows):")
print("============================\n")
print(sample_data.head(15))
print("\nNumber of Instances in Sample Data:", len(sample_data))
print("Saved: sample-data.csv")
````

### Step 5: Label Encoding the Sample Data (Input and Output is converted in Numeric Representation)

#### Step 5.1: Train the Label Encoder

As in the improved Titanic code, **MajorVessels** is a count (0 – 3), so it is encoded in its natural order (Zero=0, One=1, Two=2, Three=3)
instead of the alphabetical order `LabelEncoder` would use.

````python
"""
Purpose of this section:
-------------------------
This part of the code sets up and trains Label Encoders.
Label Encoding is used to convert categorical values (like 'Asymptomatic', 'Normal')
into numeric codes that machine learning algorithms can work with.

Steps:
1. Define sample label categories for each attribute.
2. Initialize a LabelEncoder for each attribute.
3. Train (fit) each LabelEncoder with its respective category values.
4. Encode MajorVessels with an ordinal mapping that keeps the natural order.
"""

# Define label categories for each feature
# These are possible values each column can have in the dataset
chest_pain = pd.DataFrame({"ChestPainType": ["Typical Angina", "Atypical Angina", "Non-Anginal Pain", "Asymptomatic"]})
exercise_angina = pd.DataFrame({"ExerciseAngina": ["Yes", "No"]})
thal = pd.DataFrame({"Thal": ["Normal", "Fixed Defect", "Reversible Defect"]})
heart_disease = pd.DataFrame({"HeartDisease": ["Yes", "No"]})

# Create LabelEncoder objects for each categorical column
chest_pain_label_encoder = LabelEncoder()
exercise_angina_label_encoder = LabelEncoder()
thal_label_encoder = LabelEncoder()
heart_disease_label_encoder = LabelEncoder()

# Train (fit) the encoders on the given categories
# np.ravel() flattens the DataFrame column into a 1D array for the encoder
chest_pain_label_encoder.fit(np.ravel(chest_pain))
exercise_angina_label_encoder.fit(np.ravel(exercise_angina))
thal_label_encoder.fit(np.ravel(thal))
heart_disease_label_encoder.fit(np.ravel(heart_disease))

# MajorVessels is a count, so keep its natural order (Zero < One < Two < Three)
major_vessels_ordinal_mapping = {"Zero": 0, "One": 1, "Two": 2, "Three": 3}

# Show the code assigned to every category
print("ChestPainType  :", dict(zip(chest_pain_label_encoder.classes_, chest_pain_label_encoder.transform(chest_pain_label_encoder.classes_))))
print("ExerciseAngina :", dict(zip(exercise_angina_label_encoder.classes_, exercise_angina_label_encoder.transform(exercise_angina_label_encoder.classes_))))
print("MajorVessels   :", major_vessels_ordinal_mapping)
print("Thal           :", dict(zip(thal_label_encoder.classes_, thal_label_encoder.transform(thal_label_encoder.classes_))))
print("HeartDisease   :", dict(zip(heart_disease_label_encoder.classes_, heart_disease_label_encoder.transform(heart_disease_label_encoder.classes_))))
````

#### Step 5.2: Label Encode the Output

````python
"""
Purpose of this section:
-------------------------
This part of the code applies Label Encoding to the target/output variable
("HeartDisease") in the dataset. The categorical values ("Yes", "No") are
converted into numerical values (1 and 0) that can be used by
machine learning algorithms.
"""

# Create copies of the dataset (one for encoded output, one to keep original)
sample_data_encoded_output = sample_data.copy()
original_sample_data = sample_data.copy()

# Transform "HeartDisease" (categorical -> numeric) using the trained LabelEncoder
print("\n\nHeartDisease Attribute After Label Encoding:")
print("============================================\n")
sample_data["encoded_heart_disease"] = heart_disease_label_encoder.transform(sample_data["HeartDisease"])

# Show original "HeartDisease" values and their numeric encodings
print(sample_data[["HeartDisease", "encoded_heart_disease"]].head(10))

# Replace original categorical output column with encoded values for the output dataset
sample_data_encoded_output["HeartDisease"] = sample_data["encoded_heart_disease"]

# Save the encoded dataset to a CSV file (useful for training/testing later)
sample_data_encoded_output.to_csv(os.path.join(PROJECT_DIR, "sample-data-encoded-output.csv"), index=False, header=True)
print("\nSaved: sample-data-encoded-output.csv")
````

#### Step 5.3: Label Encode the Input

````python
"""
Purpose of this section:
-------------------------
This part of the code applies Label Encoding to the input features
("ChestPainType", "ExerciseAngina", "MajorVessels", "Thal"). Each categorical
value is converted into a numeric value so that machine learning models
can process them.
"""

# Create copies of the dataset (to keep original values safe)
sample_data_encoded = sample_data_encoded_output.copy()

# Encode each input attribute
sample_data_encoded["ChestPainType"] = chest_pain_label_encoder.transform(sample_data_encoded_output["ChestPainType"])
sample_data_encoded["ExerciseAngina"] = exercise_angina_label_encoder.transform(sample_data_encoded_output["ExerciseAngina"])
sample_data_encoded["MajorVessels"] = sample_data_encoded_output["MajorVessels"].map(major_vessels_ordinal_mapping)
sample_data_encoded["Thal"] = thal_label_encoder.transform(sample_data_encoded_output["Thal"])

# Print encoded dataset (with numeric values instead of categories)
print("\n\nSample Data after Label Encoding (first 15 rows):")
print("=================================================\n")
print(sample_data_encoded.head(15))

# Save the encoded dataset to a new CSV file
sample_data_encoded.to_csv(os.path.join(PROJECT_DIR, "sample-data-encoded.csv"), index=False, header=True)
````

### Step 6: Execute the Training Phase

#### Step 6.1: Splitting Sample Data into Training Data and Testing Data

````python
"""
Purpose of this section:
-------------------------
This part of the code splits the dataset into Training Data and Testing Data.
- Training Data is used to teach the machine learning model.
- Testing Data is used to check how well the model performs on unseen data.
The rows are shuffled, and stratify keeps the same Yes/No ratio in both parts.
"""

# Split dataset into training (80%) and testing (20%)
# random_state=0 ensures reproducibility (same split every time)
training_data_encoded, testing_data_encoded = train_test_split(
    sample_data_encoded,
    test_size=0.2,
    random_state=0,
    shuffle=True,
    stratify=sample_data_encoded["HeartDisease"]
)

# Save the training and testing datasets to CSV files
training_data_encoded.to_csv(os.path.join(PROJECT_DIR, "training-data-encoded.csv"), index=False, header=True)
testing_data_encoded.to_csv(os.path.join(PROJECT_DIR, "testing-data-encoded.csv"), index=False, header=True)

print("Training Data shape:", training_data_encoded.shape, "| HeartDisease counts:", training_data_encoded["HeartDisease"].value_counts().to_dict())
print("Testing Data shape :", testing_data_encoded.shape, "| HeartDisease counts:", testing_data_encoded["HeartDisease"].value_counts().to_dict())
````

#### Step 6.2: Splitting Input Vectors and Outputs / Labels of Training Data

````python
"""
Purpose of this section:
-------------------------
This part of the code separates the features (input vectors) and
the labels (output values) from the training dataset.
"""

# Extract input vectors (features) from training data
input_vector_train = training_data_encoded.iloc[:, :-1]

# Extract output labels (target variable) from training data
output_label_train = training_data_encoded.iloc[:, -1]

print("Input vectors (train) shape:", input_vector_train.shape)
print("Output labels (train) shape:", output_label_train.shape)
````

#### Step 6.3: Train the Support Vector Classifier

As in the improved Titanic code, each SVC kernel is scored with **5-fold cross-validation on the training data only**, and the kernel
with the highest mean accuracy is used.

````python
"""
Purpose of this section:
-------------------------
This part of the code trains a Support Vector Classifier (SVC) on the training data.
The SVC is a type of machine learning model that finds the best boundary
to separate different classes (heart disease vs. no heart disease).
The kernel is selected with 5-fold cross-validation on the training data.
"""

print("\n\nChoosing the SVC Kernel with 5-Fold Cross-Validation on Training Data")
print("=====================================================================\n")

cv_folds = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)
kernel_table = PrettyTable(["Kernel", "Mean CV Accuracy"])
kernel_scores = {}
for kernel in ["linear", "rbf", "poly", "sigmoid"]:
    scores = cross_val_score(svm.SVC(kernel=kernel, gamma="auto", random_state=0),
                             input_vector_train, np.ravel(output_label_train), cv=cv_folds)
    kernel_scores[kernel] = scores.mean()
    kernel_table.add_row([kernel, round(scores.mean(), 3)])
print(kernel_table)

best_kernel = max(kernel_scores, key=kernel_scores.get)
print("\nSelected kernel:", best_kernel)

print("\n\nTraining the Support Vector Classifier on Training Data")
print("========================================================\n")

# Initialize the Support Vector Classifier with the selected kernel
svc_model = svm.SVC(kernel=best_kernel, gamma="auto", random_state=0)

# Train the model using the training data (features and labels)
svc_model.fit(input_vector_train, np.ravel(output_label_train))

# Print trained model details (parameters and values)
print(svc_model)
````

#### Step 6.4: Save the Trained Model

````python
"""
Purpose of this section:
-------------------------
This part of the code saves the trained Support Vector Classifier (SVC)
model to your computer so it can be reused later without retraining.
The same file is loaded by the Streamlit web app (app.py).
"""

# Save the trained SVC model to a file named "svc_trained_model.pkl"
with open(os.path.join(PROJECT_DIR, "svc_trained_model.pkl"), "wb") as model_file:
    pickle.dump(svc_model, model_file)
print("Saved: svc_trained_model.pkl")
````

### Step 7: Execute the Testing Phase

#### Step 7.1: Splitting Input Vectors and Outputs/Labels of Testing Data

````python
"""
Purpose of this section:
-------------------------
This part of the code separates the features (inputs) and the labels (outputs)
from the testing dataset.
"""

input_vector_test = testing_data_encoded.iloc[:, :-1]
output_label_test = testing_data_encoded.iloc[:, -1]

print("Input vectors (test) shape:", input_vector_test.shape)
print("Output labels (test) shape:", output_label_test.shape)
````

#### Step 7.2: Load the Saved Model

````python
"""
Purpose of this section:
-------------------------
This part of the code loads the previously saved Support Vector Classifier (SVC)
model from disk into memory.
"""

with open(os.path.join(PROJECT_DIR, "svc_trained_model.pkl"), "rb") as model_file:
    model = pickle.load(model_file)
print("Model loaded from svc_trained_model.pkl")
````

#### Step 7.3: Evaluate the Machine Learning Model

##### Step 7.3.1: Make Predictions with the Trained Models on Testing Data

````python
"""
Purpose of this section:
-------------------------
This part of the code evaluates the trained Support Vector Classifier (SVC)
on the testing data.
"""

model_predictions = model.predict(input_vector_test)

testing_data_encoded = testing_data_encoded.copy()
testing_data_encoded["Predictions"] = model_predictions
testing_data_encoded.to_csv(os.path.join(PROJECT_DIR, "model-predictions.csv"), index=False, header=True)

model_predictions = testing_data_encoded
print("\n\nPredictions Returned by svc_trained_model:")
print("==========================================\n")
print(model_predictions)
````

#### Step 7.4: Calculate the Accuracy Score

````python
"""
Purpose of this section:
-------------------------
This part of the code calculates how accurate the trained model is on the
testing dataset.
"""

model_accuracy_score = accuracy_score(
    model_predictions["HeartDisease"],
    model_predictions["Predictions"]
)

print("\n\nAccuracy Score:")
print("===============\n")
print(round(model_accuracy_score, 2))
````

#### Step 7.5: Detailed Evaluation Metrics (Precision, Recall, F1, Confusion Matrix)

````python
"""
Purpose of this section:
-------------------------
Accuracy alone can be misleading. This section adds the standard classification metrics:
- Precision: of everyone the model predicted as "Heart Disease", how many actually have it?
- Recall: of everyone who actually has heart disease, how many did the model catch?
  (In medicine, recall is especially important: a missed patient is a costly error.)
- F1-score: the harmonic mean of precision and recall.
- Confusion Matrix: a table of correct vs. incorrect predictions per class.
"""

print("\nClassification Report:")
print("=======================\n")
print(classification_report(
    model_predictions["HeartDisease"],
    model_predictions["Predictions"],
    target_names=["No Heart Disease (0)", "Heart Disease (1)"]
))

conf_matrix = confusion_matrix(model_predictions["HeartDisease"], model_predictions["Predictions"])
conf_matrix_df = pd.DataFrame(
    conf_matrix,
    index=["Actual: No Disease", "Actual: Disease"],
    columns=["Predicted: No Disease", "Predicted: Disease"]
)
print("\nConfusion Matrix:")
print("==================\n")
print(conf_matrix_df)

plt.figure(figsize=(5, 4))
sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Reds",
            xticklabels=["No Disease", "Disease"],
            yticklabels=["No Disease", "Disease"])
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix - SVC on Heart Disease Testing Data")
plt.tight_layout()
plt.show()
````

#### Step 7.6: Cross-Validated Accuracy

The testing data has 60 rows, so one wrong prediction changes accuracy by about 1.7%. To get a more reliable estimate, the model is also
scored with **repeated stratified 5-fold cross-validation** (5 folds × 10 repeats = 50 rounds) on all 297 rows. For reference, the setup of the
original (un-improved) Titanic code (alphabetical encoding of the count attribute + default RBF kernel) is scored on the same folds.

````python
"""
Purpose of this section:
-------------------------
Score the final setup with repeated stratified 5-fold cross-validation and compare it
with the setup of the original Titanic code (alphabetical encoding + default RBF SVC).
"""

# Original-code style encoding: LabelEncoder on MajorVessels (alphabetical: One=0, Three=1, Two=2, Zero=3)
original_major_vessels_label_encoder = LabelEncoder().fit(["Zero", "One", "Two", "Three"])
input_original = sample_data_encoded.iloc[:, :-1].copy()
input_original["MajorVessels"] = original_major_vessels_label_encoder.transform(original_sample_data["MajorVessels"])

input_final = sample_data_encoded.iloc[:, :-1]
output_all = sample_data_encoded.iloc[:, -1]

repeated_cv = RepeatedStratifiedKFold(n_splits=5, n_repeats=10, random_state=0)
original_cv_scores = cross_val_score(svm.SVC(gamma="auto", random_state=0), input_original, output_all, cv=repeated_cv)
final_cv_scores = cross_val_score(svm.SVC(kernel=best_kernel, gamma="auto", random_state=0), input_final, output_all, cv=repeated_cv)

comparison_table = PrettyTable(["Setup", "MajorVessels Encoding", "SVC Kernel", "Mean CV Accuracy", "Std"])
comparison_table.add_row(["Original Titanic code setup", "Alphabetical", "rbf", round(original_cv_scores.mean(), 3), round(original_cv_scores.std(), 3)])
comparison_table.add_row(["Final (this notebook)", "Ordinal", best_kernel, round(final_cv_scores.mean(), 3), round(final_cv_scores.std(), 3)])
print("\nRepeated Stratified 5-Fold Cross-Validation (50 rounds):")
print(comparison_table)
````
